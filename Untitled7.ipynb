3.Регресионный анализ

```python
x = df.drop('AveRooms',axis = 1)
y = df['AveRooms']
```

убираем целевую переменую чтобы модель училась предсказывать её

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

данные разделены на train/test в соотнощении 80/20

перед построением моделей посмотрим не вознкает ли линейной зависисмотси между регрессорами, если да то исключаем такие переменные

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

#Создаем DataFrame с константой для VIF
X_vif = sm.add_constant(x_train)

#Считаем VIF для каждого признака
vif_data = pd.DataFrame()
vif_data["Feature"] = X_vif.columns
vif_data["VIF"] = [variance_inflation_factor(X_vif.values, i)
                   for i in range(X_vif.shape[1])]

print(vif_data.sort_values("VIF", ascending=False))
```

два регрессора  Latitude и Longtitude сильно кореелируют между собой, поэтому исключаем их из будущих моделей

```python
for_drop = ['Latitude', 'Longitude']
x_train_reduced = x_train.drop(for_drop, axis=1)
x_test_reduced = x_test.drop(for_drop, axis=1)

```

```python
x_vif_new = sm.add_constant(x_train_reduced)

vif_data_new = pd.DataFrame()
vif_data_new["Feature"] = x_vif_new.columns
vif_data_new["VIF"] = [variance_inflation_factor(x_vif_new.values, i)
                   for i in range(x_vif_new.shape[1])]

print(vif_data_new.sort_values("VIF", ascending=False))
```

значения VIF у всез переменных < 5 ,можно приступить к обучение модели

```python
x_train_const = sm.add_constant(x_train_reduced)
model_ols = sm.OLS(y_train, x_train_const).fit()
print(model_ols.summary())
```

AveBedrms (coef 3.93): Самый сильный предиктор. Увеличение числа спален на 1 ведет к росту общего числа комнат на ~3.93.  MedInc (coef 0.48): Рост дохода увеличивает размер жилья.  HouseAge (coef -0.0108): Более старые дома имеют чуть меньше комнат (отрицательный коэффициент, p-value < 0.05). Все признаки статистически значимы.

Для получения предсказаний y_pred_clean и оценки точности модели на тестовых данных использовался класс LinearRegression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

lr_model = LinearRegression()
lr_model.fit(x_train_reduced, y_train)
y_pred_clean = lr_model.predict(x_test_reduced)

r2_clean = r2_score(y_test, y_pred_clean)
rmse_clean = np.sqrt(mean_squared_error(y_test, y_pred_clean))

print(f"R2 (без гео-признаков): {r2_clean:.2f}")
print(f"RMSE (без гео-признаков): {rmse_clean:.2f}")
```

визуализируем регресиию

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred_clean, alpha=0.5, color='blue')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r', lw=2, label="Идеальное предсказание")
plt.xlabel('Фактическое AveRooms')
plt.ylabel('Предсказанное AveRooms')
plt.title('Регрессия: Фактически и Предсказательно')
plt.legend()
plt.grid(True)
plt.show()
```

Визуальное разделение точек на графике рассеяния связано со спецификой распределения жилого фонда в данных:Основное облако AveRooms  Соответствует 99% всех районных блоков - типовой многоквартирной и частной застройке городов, где модель работает с минимальной ошибкой.  Одиночные точки AveRooms > 10.0: Отражают редкие малонаселенные районы с крупногабаритными объектами и загородными домовладениями. Высокий коэффициент при AveBedrms (3.93) экстраполирует модель на эти единичные точки, создавая видимость визуального «хвоста».

4.Классификация

```python
median_val = df['AveRooms'].median()
print("Медиана AveRooms:",median_val)
```

```python
df['AveRooms_Binary'] = (df['AveRooms'] > median_val).astype(int)
```

Создан бинарный целевой признак AveRooms_Binary на основе медианы (5.22):
0(Мало комнат):AveRooms <= 5.22
1(Много комнат):AveRooms > 5.22


```python
for_drop_class = ['AveRooms', 'AveRooms_Binary', 'Latitude', 'Longitude']
x_class = df.drop(columns=for_drop_class)
y_class = df['AveRooms_Binary']
```

Из матрицы признаков исключен исходный AveRooms( и AveRooms_Binary) во избежание утечки данных


```python
# Разделение выборки
x_train_lr, x_test_lr, y_train_lr, y_test_lr = train_test_split(
    x_class, y_class, test_size=0.2, random_state=42
)
```

масштабриуем данные(и тренировочные и тестовые) это важно для логистической регрессии

```python
from sklearn.preprocessing import StandardScaler
scaler_lr = StandardScaler()
x_train_lr_scaled = scaler_lr.fit_transform(x_train_lr)
x_test_lr_scaled = scaler_lr.transform(x_test_lr)
```

модель 1: Логистическая регрессия

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
logistic_model = LogisticRegression(random_state=42,max_iter=1000)
logistic_model.fit(x_train_lr_scaled,y_train_lr)
y_prediction_logistic = logistic_model.predict(x_test_lr_scaled)
print(classification_report(y_test_lr,y_prediction_logistic,target_names = ["Мало Комнат","Много Комнат"]))
```

Модель показала общую точность (Accuracy)~ 0.80%, что является хорошим базовым результатом для бинарной классификации на основе социально-экономических признаков. у recall тоже неплохой показатель модель пропускает( в среднем) 20% блоков , f1-score ~80% говорит о том что модель сбалансированная и не перекошена в сторону одного из классов


модель 2: случайный лес

```python
from sklearn.ensemble import RandomForestClassifier
forest_model = RandomForestClassifier(n_estimators=100, random_state = 42)
forest_model.fit(x_train_lr,y_train_lr)
y_pred_forest = forest_model.predict(x_test_lr)
y_pred_forest
print(classification_report(y_test_lr,y_pred_forest,target_names = ["Мало Комнат","Много Комнат"]))
```

модель стала показывать метрики на 4% лучше по всем показателям, это говорит о наличие нелинейных связей между данными которая линейная регрессия не смогла уловить

важность признаков для слусайного леса

```python
importances = forest_model.feature_importances_
feat_imp = pd.Series(importances, index = x_class.columns).sort_values(ascending=False)

plt.figure(figsize=(8, 5))
feat_imp.plot(kind='bar', color='teal')
plt.title('Важность признаков (Random Forest)')
plt.ylabel('Importance')
plt.xlabel('Feature')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Матрица ошибок позволяет увидеть не просто общую точность, а типичные ошибки модели. Рассмотрим матрицы для обеих моделей

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm = confusion_matrix(y_test_lr,y_pred_forest)
disp = ConfusionMatrixDisplay(cm,display_labels=["Мало комнат", "Много комнат"])
plt.figure(figsize=(6, 5))
disp.plot(cmap='Blues', values_format='d')
plt.title('Матрица ошибок (Random Forest)')
plt.show()
```

Интерпретация:

True Negatives (1838):Модель верно идентифицировала большинство блоков с малым количеством комнат.

True Positives (1716): Модель верно нашла ~83% многокомнатных блоков.

False Positives (251 — Ошибка I рода): Модель назвала 251 блоков многокомнатными, хотя там мало комнат. Это означает, что модель иногда переоценивает размер жилья (возможно, в районах с высоким доходом, но компактными квартирами).

False Negatives (310 — Ошибка II рода)* Модель пропустила 310 многокомнатных блоков, отнеся их к классу «мало». Это самая частая ошибка модели.


```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm = confusion_matrix(y_test_lr,y_prediction_logistic)
disp = ConfusionMatrixDisplay(cm,display_labels=["Мало комнат", "Много комнат"])
plt.figure(figsize=(6, 5))
disp.plot(cmap='Blues', values_format='d')
plt.title('Матрица ошибок (логистическая регресиия)')
plt.show()
```

У логистической регрессии ошибок типа False Negative больше, то есть она пропускает около 397 многокомнатных домов. Random Forest справляется лучше именно за счет снижения количества пропусков (False Negatives), что делает его более надежным для поиска просторного жилья.


Вывод по матрице:
Модель склонна немного занижать количество комнат (ошибок FN > FP). Это согласуется с природой данных: блоки со значением, близким к медиане (5.22), сложно однозначно отнести к классу «Много», и модель перестраховывается, относя их к «Мало».

5.Кластеризация

```python
df = df.drop('AveRooms_Binary',axis = 1)
numeric_cols=df.select_dtypes(include=["int64","float64"]).columns
print("\n5. Числовые признаки:")
numeric_cols
```

выбираем все числовые признаки, для кластеризации методом k-means

```python
df_clust = df.drop(columns=['AveRooms_Binary'], errors='ignore').copy()
```

```python
#Фильтрация экстремальных аномалий по AveOccup (например, > 10) и AveBedrms (например, > 5)
df_clust = df_clust[(df_clust['AveOccup'] < 10) & (df_clust['AveBedrms'] < 5)]
```

```python
numeric_cols = df_clust.select_dtypes(include=["int64", "float64"]).columns
x_cluster = df_clust[numeric_cols].copy()
```

стандартизация

```python
scaler_clust = StandardScaler()
x_scaled = scaler_clust.fit_transform(x_cluster)
```

Данные были стандартизированы (StandardScaler),так как алгоритм K-Means чувствителен к разным единицам измерения (например, население в тысячах, а комнаты - в единицах)


```python
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score


# Диапазон проверки числа кластеров
k_range = range(2, 9)
silhouette_scores = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(x_scaled)

    # Расчет среднего силуэта
    score = silhouette_score(x_scaled, cluster_labels)
    silhouette_scores.append(score)
    print(f'k = {k}: Silhouette Score = {score:.4f}')

# Построение графика
plt.figure(figsize=(8, 4))
plt.plot(k_range, silhouette_scores, 'bo-', linewidth=2)
plt.xlabel('Количество кластеров k')
plt.ylabel('Средний коэффициент силуэта')
plt.title('Определение k методом силуэта')
plt.grid(True)
plt.show()

```

Методом силуета была выведено что оптимальное количество кластеров для наших данных равна 3

подготавливаем данные для визуализации, обучаем модель кластеризации и снижаем размерность

```python
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(x_scaled)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(x_scaled)

pca_centroids = pca.transform(kmeans.cluster_centers_)
```

Визуализация кластеров

```python
plt.figure(figsize=(9, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='Set2', s=25, alpha=0.6)
plt.scatter(pca_centroids[:, 0], pca_centroids[:, 1], marker='X', s=200, c='red', edgecolor='black', label='Центроиды')
plt.title('Визуализация кластеров K-Means через PCA')
plt.xlabel('Главная компонента 1 (PC1)')
plt.ylabel('Главная компонента 2 (PC2)')
plt.colorbar(scatter, label='Номер кластера')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

После удаления аномалий AveRooms < 20 проекция кластеров в пространстве PCA продемонстрировала отсутствие сильного искажения масштаба осей. Основная масса объектов сформировала два плотных ядра (компактная и просторная застройка), а объекты с высокими значениями комнатности выделены K-Means в обособленный верхний кластер
